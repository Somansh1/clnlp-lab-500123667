```python
import re, string

passage = (
    "Natural Language Processing (NLP) enables machines to read, parse, "
    "and generate human text! In 2026, enterprise pipelines process over "
    "10,000 documents per minute across 40+ languages. However, raw corpora "
    "contain noisy punctuation, mixed casing, and frequent function words. "
    "Tokenization splits continuous text into discrete sentences and words. "
    "Stop-word filtering removes terms such as 'the', 'is', and 'and'. "
    "Clean tokens improve downstream TF-IDF and neural models. "
    "Careful preprocessing reduces vocabulary size by 20-35%. "
    "Every robust CLNLP workflow begins with reproducible text cleaning."
)
print("Upper:", sum(1 for c in passage if c.isupper()),
      "Punct:", sum(1 for c in passage if c in string.punctuation),
      "Digits:", sum(1 for c in passage if c.isdigit()),
      "Spaces:", sum(1 for c in passage if c.isspace()))
sents = [s.strip() for s in re.split(r"(?<=[.!?])\s+", passage) if s.strip()]
tokens = re.findall(r"\b[A-Za-z]+\b", passage.lower())
sw = {"in","to","and","of","the","is","are","as","by","with","into","over","per","such","every","across","however"}
filtered = [t for t in tokens if t not in sw]
print(f"Sentences: {len(sents)} | Tokens: {len(tokens)} | Filtered: {len(filtered)}")
```
