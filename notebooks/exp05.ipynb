```python
from collections import Counter

corpus = [
    "natural language processing and deep learning models",
    "language models learn subword representations for machine learning",
    "byte pair encoding merges frequent character pairs in natural language"
]
vocab = Counter()
for line in corpus:
    for w in line.lower().split():
        vocab[" ".join(list(w)) + " </w>"] += 1

merges = []
for _ in range(8):
    pairs = Counter()
    for w, f in vocab.items():
        syms = w.split()
        for i in range(len(syms) - 1):
            pairs[(syms[i], syms[i + 1])] += f
    best = max(pairs, key=pairs.get)
    merges.append("".join(best))
    bi = " ".join(best)
    vocab = Counter({w.replace(bi, "".join(best)): f for w, f in vocab.items()})
print("Top 8 BPE merges:", merges)
```
