```python
import re

sample = (
    "Contact coordinator@upes.ac.in or somansh.goel@gmail.com for CLNLP lab. "
    "Visit https://www.upes.ac.in and https://github.com/Somansh1. "
    "Call +91-9876543210 or 9123456789. Follow @UPESDehradun and "
    "@AI_RoboticsLab #CLNLP #Python #MachineLearning"
)
print("Emails:", re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", sample))
print("URLs:", re.findall(r"https?://[^\s]+", sample))
print("Phones:", re.findall(r"(?:\+91[- ]?)?[6-9]\d{9}", sample))
print("Hashtags:", re.findall(r"#[A-Za-z0-9_]+", sample))
print("Mentions (naive):", re.findall(r"@[A-Za-z0-9_]+", sample))
print("Mentions (fixed):", re.findall(r"(?<!\S)@[A-Za-z0-9_]+", sample))
```
