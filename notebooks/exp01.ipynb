```python
import csv, statistics
from collections import Counter, defaultdict

with open("../data/employee_information_100.csv", newline="", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

salaries = [float(r["Salary"]) for r in rows]
mean_sal = statistics.mean(salaries)
above_avg = sum(1 for s in salaries if s > mean_sal)
print(f"Records: {len(rows)} | Mean Salary: {mean_sal:.1f} | Max: {max(salaries):.0f} | Above Mean: {above_avg}")
print("Department counts:", dict(Counter(r["Department"] for r in rows)))
print("Gender counts:", dict(Counter(r["Gender"] for r in rows)))
```
