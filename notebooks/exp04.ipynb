# Experiment 4: Term Frequency, NER & Manual TF-IDF

First-principles smoothed TF-IDF calculation across a 3-document corpus.

```python
import math, re
from collections import Counter

docs = [
    "Natural language processing is a field of artificial intelligence",
    "Natural language processing helps computers understand human language",
    "Machine learning is an important part of artificial intelligence"
]
toks = [re.findall(r"\b[a-z]+\b", d.lower()) for d in docs]
vocab = sorted({w for d in toks for w in d})
N = len(docs)
df = {w: sum(1 for d in toks if w in d) for w in vocab}
for idx, d in enumerate(toks, 1):
    cnt = Counter(d)
    scores = sorted(((w, round((cnt[w]/len(d))*(math.log((N+1)/(df[w]+1))+1.0), 4)) for w in cnt),
                    key=lambda x: (-x[1], x[0]))[:5]
    print(f"Doc {idx} top TF-IDF:", scores)
```
